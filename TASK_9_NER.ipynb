```python
import spacy

# Load the English model
nlp = spacy.load("en_core_web_sm")
```

```python
text = "Elon Musk founded SpaceX and Tesla in the United States."
doc = nlp(text)
```

```python
print("Named Entities:\n")

for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```

```python
from spacy import displacy

displacy.render(doc, style="ent", jupyter=True)
```

1000 dataset code

```python
# Step 1: Import libraries
import random
import pandas as pd

# Step 2: Sample data for sentence generation
persons = ["Elon Musk", "Bill Gates", "Virat Kohli", "Sundar Pichai", "Narendra Modi"]
organizations = ["Google", "Microsoft", "Tesla", "BCCI", "ISRO"]
locations = ["India", "USA", "Hyderabad", "Delhi", "Bangalore"]
verbs = ["visited", "founded", "announced", "joined", "launched", "met"]
objects = ["a new project", "a startup", "an initiative", "a program", "a company"]

# Step 3: Function to generate one sentence
def generate_sentence():
    person = random.choice(persons)
    org = random.choice(organizations)
    location = random.choice(locations)
    verb = random.choice(verbs)
    obj = random.choice(objects)

    sentence = f"{person} from {org} {verb} {obj} in {location}."
    return sentence

# Step 4: Generate 1000 sentences
sentences = [generate_sentence() for _ in range(1000)]

# Step 5: Create DataFrame
df = pd.DataFrame({"Sentence": sentences})

# Step 6: Save dataset
df.to_csv("ner_dataset_1000_sentences.csv", index=False)

# Step 7: Display sample
df.head()
```

```python
def generate_labeled_sentence():
    person = random.choice(persons)
    org = random.choice(organizations)
    location = random.choice(locations)
    verb = random.choice(verbs)
    obj = random.choice(objects)

    sentence = f"{person} from {org} {verb} {obj} in {location}."

    return {
        "sentence": sentence,
        "person": person,
        "organization": org,
        "location": location
    }

data = [generate_labeled_sentence() for _ in range(1000)]
df = pd.DataFrame(data)
df.to_csv("ner_labeled_dataset.csv", index=False)

df.head()
```
