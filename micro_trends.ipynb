```python
# 1. Install necessary libraries
!pip install pandas ntscraper docx2txt

import pandas as pd
from ntscraper import Nitter # A wrapper for scraping social signals without API keys
import docx2txt

# 2. Load your Market Research to identify target cities (Mumbai, Kolkata, etc.)
# Reference: Market_Research_Report.docx
try:
    text = docx2txt.process("Market_Research_Report.docx")
    print("Market Research Loaded. Targeting top spending cities: Kolkata and Mumbai.")[cite: 1]
except:
    print("Please upload Market_Research_Report.docx to the files tab.")

# 3. Initialize the Scraper Agent
scraper = Nitter()

def get_street_signals(search_term, count=50):
    """Function to pull real-time street style mentions"""
    try:
        # We search for the 'Bambaiya' keywords you identified
        tweets = scraper.get_tweets(search_term, mode='term', number=count)
        data = []
        for tweet in tweets['tweets']:
            data.append({
                'date': tweet['date'],
                'text': tweet['text'],
                'user': tweet['user']['name'],
                'source': 'Street_Signal'
            })
        return pd.DataFrame(data)
    except Exception as e:
        print(f"Scraper notice: {e}")
        return pd.DataFrame()

# 4. Run 'Global' vs 'Local' Signal Ingestion
print("Ingesting Global High-Fashion Signals...")
# Tracking 'Industrial Streetwear' as identified in your strategy
global_signals = get_street_signals("Industrial Streetwear", count=20)

print("Ingesting Local Bambaiya Signals...")
# Using your specific 'Tod-fod' and 'Sadak ka Canvas' keywords
local_signals_1 = get_street_signals("Tod-fod fashion", count=20)
local_signals_2 = get_street_signals("Sadak ka Canvas", count=20)

# 5. Merge and Organize for Analysis
all_signals = pd.concat([global_signals, local_signals_1, local_signals_2], ignore_index=True)

# 6. Display the Signal Intersection
print("\n--- Emerging Micro-Trend Data Stream ---")
print(all_signals.head(10))

# Export for the next step: Sentiment Analysis
all_signals.to_csv('emerging_trends_raw.csv', index=False)
```

```python
# 1. Install Hugging Face Transformers for NLP
!pip install transformers scipy

from transformers import pipeline
import pandas as pd
import numpy as np

# 2. Initialize the Sentiment Analysis Agent (Agent 03: Linguistic Localizer)
# We use a model that understands Hinglish/multilingual nuances
sentiment_pipeline = pipeline("sentiment-analysis", model="cardiffnlp/twitter-xlm-roberta-base-sentiment")

# 3. Load the data from Step 1
df = pd.DataFrame() # Initialize df to an empty DataFrame to prevent NameError
try:
    temp_df = pd.read_csv('emerging_trends_raw.csv')
    if not temp_df.empty:
        df = temp_df
        print(f"Analyzing {len(df)} signals from Mumbai and Kolkata...")
    else:
        print("Warning: 'emerging_trends_raw.csv' is empty. No signals to analyze.")
except pd.errors.EmptyDataError:
    print("Error: 'emerging_trends_raw.csv' is empty or corrupt. Run Step 1 first to generate the CSV with actual data.")
except FileNotFoundError:
    print("Error: 'emerging_trends_raw.csv' not found. Run Step 1 first to generate the CSV.")
except Exception as e: # Catch any other unexpected errors
    print(f"An unexpected error occurred while reading the CSV: {e}")

if df.empty:
    print("Skipping further analysis due to no data.")
else:
    # 4. Process Sentiment for each 'Signal'
    def analyze_vibe(text):
        # Truncate text for the model
        result = sentiment_pipeline(text[:512])[0]
        # Convert labels to scores: Positive = 1, Neutral = 0, Negative = -1
        label = result['label']
        score = result['score']
        if 'Positive' in label or 'label_2' in label: return score
        if 'Negative' in label or 'label_0' in label: return -score
        return 0

    print("Calculating Trend Confidence Scores...")
    df['vibe_score'] = df['text'].apply(analyze_vibe)

    # 5. Strategic Weighting: Give more importance to top-spending cities
    # As per Market_Research_Report.docx, Kolkata and Mumbai are priority
    df['market_weight'] = df['text'].apply(lambda x: 1.5 if any(city in x.lower() for city in ['mumbai', 'kolkata']) else 1.0)
    df['final_score'] = df['vibe_score'] * df['market_weight']

    # 6. The Trend Radar Output
    trend_summary = df.groupby('source').agg({
        'final_score': 'mean',
        'text': 'count'
    }).rename(columns={'text': 'signal_volume'})

    print("\n--- AI Trend Prediction Dashboard ---")
    print(trend_summary)

    # 7. Decision Logic
    for trend, row in trend_summary.iterrows():
        if row['final_score'] > 0.5:
            print(f"🚀 ACTION: Launch 'Tod-fod' campaign in Mumbai immediately. High Confidence.")
        elif row['final_score'] < 0:
            print(f"⚠️ RISK: '{trend}' showing negative sentiment. Potential 'Vibe Exhaustion'.")
```

```python
# 1. Install OpenAI's CLIP and dependencies
!pip install ftfy regex tqdm
!pip install git+https://github.com/openai/CLIP.git

import torch
import clip
from PIL import Image
import requests
from io import BytesIO

# 2. Load the Model (The "Visual Agent")
device = "cuda" if torch.cuda.is_available() else "cpu"
model, preprocess = clip.load("ViT-B/32", device=device)

# 3. Define our "Off-White Mumbai" Visual Benchmarks
# These are the 'Quotes' the AI will use to judge an image
class_descriptions = [
    "Industrial streetwear with orange zip ties",
    "Traditional Indian formal wear",
    "Mumbai street style with oversized hoodies",
    "Corporate office attire",
    "Deconstructed fashion with bold text quotes"
]
text_tokens = clip.tokenize(class_descriptions).to(device)

# 4. Function to analyze a 'Street Image'
def analyze_visual_vibe(image_url):
    # Download and preprocess the image
    response = requests.get(image_url)

    # Check if the request was successful and content is an image
    if response.status_code != 200:
        print(f"Error: Failed to retrieve image from {image_url}. Status code: {response.status_code}")
        return None

    content_type = response.headers.get('Content-Type', '')
    if not content_type.startswith('image'):
        print(f"Error: Content at {image_url} is not an image. Content-Type: {content_type}")
        return None

    try:
        image = preprocess(Image.open(BytesIO(response.content))).unsqueeze(0).to(device)
    except Exception as e:
        print(f"Error: Could not process image from {image_url}. {e}")
        return None

    with torch.no_grad():
        # Calculate how well the image matches each description
        logits_per_image, _ = model(image, text_tokens)
        probs = logits_per_image.softmax(dim=-1).cpu().numpy()

    # Map results to our descriptions
    results = dict(zip(class_descriptions, probs[0]))
    return results

# 5. TEST: Analyzing a placeholder 'Mumbai Street' vibe
# (You can replace this URL with a real image of Mumbai street style)
test_url = "https://raw.githubusercontent.com/pytorch/hub/master/images/dog.jpg" # Using a reliable test image
vibe_results = analyze_visual_vibe(test_url)

if vibe_results:
    print("\n--- Visual Vibe Analysis ---")
    for style, score in vibe_results.items():
        print(f"{style}: {score*100:.2f}% Match")
else:
    print("Visual analysis skipped due to image retrieval/processing error.")
```

```python
import requests
from PIL import Image
from io import BytesIO
import clip
import torch

# 1. Setup the Visual Agent
device = "cuda" if torch.cuda.is_available() else "cpu"
model, preprocess = clip.load("ViT-B/32", device=device)

# 2. Updated Analysis Function with Error Handling
def analyze_visual_vibe_robust(image_url):
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
    }

    try:
        # Check 1: Request Status
        response = requests.get(image_url, headers=headers, timeout=10)
        response.raise_for_status()

        # Check 2: Content Type
        if "image" not in response.headers.get("Content-Type", ""):
            return "Error: URL does not point to a valid image."

        # Check 3: PIL Image Loading
        image_data = Image.open(BytesIO(response.content)).convert("RGB")
        image = preprocess(image_data).unsqueeze(0).to(device)

        # Define 'Off-White' Benchmarks
        labels = ["Industrial streetwear", "Luxury branding", "Mumbai street style", "Formal attire"]
        text_tokens = clip.tokenize(labels).to(device)

        with torch.no_grad():
            logits_per_image, _ = model(image, text_tokens)
            probs = logits_per_image.softmax(dim=-1).cpu().numpy()

        return dict(zip(labels, probs[0]))

    except Exception as e:
        return f"Risk Detected: {str(e)}"

# 3. Test with a reliable source
test_url = "https://raw.githubusercontent.com/pytorch/hub/master/images/dog.jpg" # Fallback test image
results = analyze_visual_vibe_robust(test_url)

print("\n--- Visual Risk & Vibe Report ---")
if isinstance(results, dict):
    for label, score in results.items():
        print(f"{label}: {score*100:.2f}% Match")
else:
    print(results)
```

```python
# ==============================
# 1. INSTALL LIBRARIES
# ==============================
!pip install transformers scipy python-docx plotly kaleido

# ==============================
# 2. IMPORT LIBRARIES
# ==============================
import pandas as pd
import numpy as np
from transformers import pipeline
import plotly.express as px
from docx import Document
from docx.shared import Inches
from google.colab import files

# ==============================
# 3. LOAD DATA
# ==============================
try:
    df = pd.read_csv('emerging_trends_raw.csv')
    df['text'] = df['text'].astype(str)
    print(f"Loaded {len(df)} records")
except:
    print("CSV not found. Creating dummy data...")
    df = pd.DataFrame({
        'text': [
            "Mumbai streetwear is trending",
            "Kolkata fashion is improving",
            "Negative feedback on pricing",
            "Mumbai youth love oversized hoodies"
        ],
        'source': ['Instagram', 'Twitter', 'Reddit', 'Instagram']
    })

# ==============================
# 4. SENTIMENT ANALYSIS
# ==============================
sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model="cardiffnlp/twitter-xlm-roberta-base-sentiment"
)

def analyze_vibe(text):
    result = sentiment_pipeline(text[:512])[0]
    label = result['label']
    score = result['score']
    if 'Positive' in label or 'label_2' in label:
        return score
    elif 'Negative' in label or 'label_0' in label:
        return -score
    else:
        return 0

df['vibe_score'] = df['text'].apply(analyze_vibe)

# ==============================
# 5. SCORING
# ==============================
df['market_weight'] = df['text'].apply(
    lambda x: 1.5 if any(city in x.lower() for city in ['mumbai', 'kolkata']) else 1.0
)
df['final_score'] = df['vibe_score'] * df['market_weight']

trend_summary = df.groupby('source').agg({
    'final_score': 'mean',
    'text': 'count'
}).rename(columns={'text': 'signal_volume'})

print("\nTrend Summary:\n", trend_summary)

# ==============================
# 6. DASHBOARD GRAPH
# ==============================
df_final = pd.DataFrame({
    "Metric": ["Linguistic Confidence", "Market Readiness", "Trend Strength"],
    "Score": [
        round(df['vibe_score'].mean()*100, 2),
        round(df['final_score'].mean()*100, 2),
        round(trend_summary['final_score'].mean()*100, 2)
    ]
})

fig = px.bar(df_final, x='Metric', y='Score', color='Metric',
             title="AI Market Insight Dashboard")

fig.show()

# Save graph
fig.write_image("dashboard.png")

# ==============================
# 7. GENERATE WORD REPORT
# ==============================
doc = Document()

doc.add_heading('AI Market Research Report', 0)

# Introduction
doc.add_heading('1. Introduction', 1)
doc.add_paragraph(
    "This report analyzes market trends using AI-based sentiment analysis and data-driven insights."
)

# Data Summary
doc.add_heading('2. Data Summary', 1)
doc.add_paragraph(f"Total Records: {len(df)}")

# Insights
doc.add_heading('3. Key Insights', 1)
doc.add_paragraph(f"Average Sentiment Score: {df['vibe_score'].mean():.2f}")
doc.add_paragraph(f"Weighted Market Score: {df['final_score'].mean():.2f}")

# Trend Table
doc.add_heading('4. Trend Analysis', 1)
table = doc.add_table(rows=1, cols=3)

hdr = table.rows[0].cells
hdr[0].text = 'Source'
hdr[1].text = 'Score'
hdr[2].text = 'Volume'

for idx, row in trend_summary.iterrows():
    cells = table.add_row().cells
    cells[0].text = str(idx)
    cells[1].text = f"{row['final_score']:.2f}"
    cells[2].text = str(row['signal_volume'])

# Graph
doc.add_heading('5. Dashboard Visualization', 1)
doc.add_picture("dashboard.png", width=Inches(5))

# Conclusion
doc.add_heading('6. Conclusion', 1)

if df['final_score'].mean() > 0.5:
    doc.add_paragraph("Strong positive trend. Recommended to launch strategy.")
else:
    doc.add_paragraph("Moderate/low trend. Further analysis required.")

# Save report
doc.save("AI_Market_Report.docx")

# ==============================
# 8. DOWNLOAD REPORT
# ==============================
files.download("AI_Market_Report.docx")
```
