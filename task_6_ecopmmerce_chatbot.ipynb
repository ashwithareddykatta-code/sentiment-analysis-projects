```python
import string
import random
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Healthcare Knowledge Base
qa_pairs = {
    "what is fever": "Fever is a temporary increase in body temperature.",
    "symptoms of fever": "High temperature, sweating, chills, headache.",
    "treatment for fever": "Drink fluids, rest, and take paracetamol if needed.",
    "what is cold": "Common cold is a viral infection.",
    "symptoms of cold": "Runny nose, sneezing, sore throat.",
    "what is diabetes": "Diabetes is high blood sugar condition.",
    "what is headache": "Pain in head region.",
    "what is covid": "COVID-19 is a respiratory illness.",
    "how to stay healthy": "Eat well, exercise, sleep properly."
}

greetings = ["hi", "hello", "hey"]

# Preprocess function
def preprocess(text):
    words = word_tokenize(text.lower())
    words = [w for w in words if w not in string.punctuation]
    words = [w for w in words if w not in stopwords.words('english')]
    return set(words)

# Single question function (IMPORTANT FIX for Colab)
def get_response(user_input):

    if user_input in greetings:
        return random.choice(["Hello!", "Hi there!", "Hey!"])

    user_words = preprocess(user_input)

    best_match = None
    max_score = 0

    for question in qa_pairs:
        question_words = preprocess(question)
        score = len(user_words.intersection(question_words))

        if score > max_score:
            max_score = score
            best_match = question

    if best_match and max_score > 0:
        return qa_pairs[best_match]
    else:
        return "Please consult a doctor for proper advice."

# Test chatbot manually (NO LOOP - WORKS IN COLAB)
print("Type your question below:")

user_input = input()
print("Chatbot:", get_response(user_input))
```

```python
import streamlit as st
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import nltk

nltk.download('punkt')
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

# E-commerce chatbot data
qa_pairs = {
    "show me shoes": "Here are some popular shoes: Nike, Adidas, Puma.",
    "show me mobiles": "Available mobiles: iPhone, Samsung, OnePlus.",
    "track my order": "Please enter your order ID to track your order.",
    "return policy": "You can return products within 7 days.",
    "payment methods": "We accept UPI, Credit Card, Debit Card.",
    "delivery time": "Delivery takes 3-5 days.",
    "hi": "Hello! Welcome to our store.",
    "hello": "Hi! How can I help you?"
}

def preprocess(text):
    words = word_tokenize(text.lower())
    words = [w for w in words if w not in string.punctuation]
    words = [w for w in words if w not in stop_words]
    return set(words)

def get_response(user_input):
    user_words = preprocess(user_input)
    best_match = None
    max_score = 0

    for question in qa_pairs:
        question_words = preprocess(question)
        score = len(user_words.intersection(question_words))

        if score > max_score:
            max_score = score
            best_match = question

    if best_match and max_score > 0:
        return qa_pairs[best_match]
    else:
        return "Sorry, I didn’t understand."

# Streamlit UI
st.title("🛒 E-Commerce Chatbot")

user_input = st.text_input("Ask your question:")

if user_input:
    response = get_response(user_input)
    st.write("🤖 Chatbot:", response)
```
